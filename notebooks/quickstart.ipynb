## 🔬 3. Calculate Empirical Hausdorff / Box-Counting Dimension
Compute the Minkowski-Bouligand fractal dimension $D = \lim_{\epsilon \to 0} \frac{\log N(\epsilon)}{\log(1/\epsilon)}$ of the **Sierpinski Triangle**:
* Theoretical: $\frac{\log 3}{\log 2} \approx 1.5850$

```python
import numpy as np
import matplotlib.pyplot as plt
from fantasticfractals import FractalEngine, compute_box_dimension

engine = FractalEngine()
grid = engine.render_raw("sierpinski_triangle", width=1024, height=1024, max_iter=20)
dim, r2, scales, counts = compute_box_dimension(grid)

print(f"Calculated Box Dimension: {dim:.4f}")
print(f"Goodness of fit (R²):     {r2:.4f}")

# Plot the log-log box counting slope
log_inv_eps = np.log(1.0 / np.array(scales))
log_n = np.log(counts)

plt.figure(figsize=(7, 4))
plt.plot(log_inv_eps, log_n, 'o-', color='#6366f1', label=f'Dimension D = {dim:.3f} (R²={r2:.3f})')
plt.title("Minkowski-Bouligand Box Counting Dimension")
plt.xlabel("log(1 / epsilon)")
plt.ylabel("log(N(epsilon))")
plt.grid(True, alpha=0.3)
plt.legend()
plt.show()
```
